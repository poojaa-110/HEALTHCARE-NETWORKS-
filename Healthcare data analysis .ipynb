DATA ANALYSIS:

1. Which medical condition affects the most patient across the network?

```python
counts=df['Medical Condition'].value_counts()
plt.figure(figsize=(12,8))
cn=sns.barplot(x=counts.index,y=counts.values)
cn.bar_label(cn.containers[0])
plt.title("Medical Condition Distribution")
plt.xlabel("Medical Condition")
plt.ylabel('No. of Patients')
plt.xticks(rotation=75)
plt.tight_layout()
plt.show()
```

From the above Visualization using bar plot we can conclude that Arthritis is the medical condition affect most patient across the network.

2.How is patient age distributed across the network?

```python

plt.figure(figsize=(8,6))
plt.hist(df['Age'],bins=40,color='Cyan', edgecolor='black')
plt.title("Age Distribution of Patients")
plt.xlabel("Age")
plt.ylabel('No. of Patients')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

From above visualization we can conclude that equally distribution of patient in all age groups.

3.How has average treatment cost change over time??

```python
average_cost=df['Billing Amount'].mean()
print(average_cost)
```

```python
df['Admission Month']=df['Date of Admission'].dt.to_period('M') # to extract month
```

```python
Monthly_average_cost=df.groupby('Admission Month')['Billing Amount'].mean()
print(Monthly_average_cost)
```

```python
plt.figure(figsize=(18,8))
plt.plot(Monthly_average_cost.index.astype(str),Monthly_average_cost.values,color='Red',marker="o",markersize='5')
plt.title("Average Treatment Cost by Month")
plt.xlabel("Month")
plt.ylabel('Avg Billing Amount')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

Above chart shows that there is stability in monthly distribution of average billing cost in overtime.

```python
df['Admission Year']=df['Date of Admission'].dt.year # to extract year
```

```python
Average_yearly_cost=df.groupby('Admission Year')['Billing Amount'].mean()
print(Average_yearly_cost)
```

```python
plt.figure(figsize=(12,6))
plt.plot(Average_yearly_cost.index.astype(str),Average_yearly_cost.values,color='Red',marker='o',markersize='5')
plt.title('Average Treatment Cost by Year')
plt.xlabel('Years')
plt.ylabel('Average billing cost')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

Above chart shows there is least average billing cost in 2020 and 2024 but highiest in 2019.

```python
Average_staying_time=df.groupby('Admission Year')['Staying time'].mean()
print(Average_staying_time)
```

Staying time has no any effect on Avg billing cost because avg staying time is similar in all years.

4. Is there any relationship between patient age and treatment cost??

```python
Sample_df=df.sample(1000,random_state=1)
plt.figure(figsize=(14,8))
plt.scatter(Sample_df['Age'],Sample_df['Billing Amount'],color='Purple',alpha=0.5,s=15)
plt.title("Age vs Treatment cost")
plt.xlabel('Age')
plt.ylabel('Treatment cost')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

Above chart shows there is no any effect of patient age on treatment cost it is equally distributed in all age patient.
